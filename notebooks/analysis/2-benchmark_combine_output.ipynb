```python
#dl2_path = "/hai/scratch/zwefers/seq2loc/deeploc2/outputs/seq2locbench/prott5"
dl2_path = f"{DEEPLOC2_OUTPUT_DIR}/prott5"


if SINGLE:
    tag="_single"
elif MULTI:
    tag="_multi"
else:
    tag = ""

test_avg_metrics = pd.read_csv(f"{dl2_path}/test_metrics_avg{tag}.csv")
test_avg_metrics["model"] = "DeepLoc2-ProtT5"
test_avgs.append(test_avg_metrics)
test_perclass_metrics_level1 = pd.read_csv(f"{dl2_path}/test_metrics_perclass_level1{tag}.csv")
test_perclass_metrics_level1["model"] = "DeepLoc2-ProtT5"
test_perclass_level1.append(test_perclass_metrics_level1)
test_perclass_metrics_level2 = pd.read_csv(f"{dl2_path}/test_metrics_perclass_level2{tag}.csv")
test_perclass_metrics_level2["model"] = "DeepLoc2-ProtT5"
test_perclass_level2.append(test_perclass_metrics_level2)
test_perclass_metrics_level3 = pd.read_csv(f"{dl2_path}/test_metrics_perclass_level3{tag}.csv")
test_perclass_metrics_level3["model"] = "DeepLoc2-ProtT5"
test_perclass_level3.append(test_perclass_metrics_level3)

val_avg_metrics = pd.read_csv(f"{dl2_path}/val_metrics_avg{tag}.csv")
val_avg_metrics["model"] = "DeepLoc2-ProtT5"
val_avgs.append(val_avg_metrics)
val_perclass_metrics_level1 = pd.read_csv(f"{dl2_path}/val_metrics_perclass_level1{tag}.csv")
val_perclass_metrics_level1["model"] = "DeepLoc2-ProtT5"
val_perclass_level1.append(val_perclass_metrics_level1)
val_perclass_metrics_level2 = pd.read_csv(f"{dl2_path}/val_metrics_perclass_level2{tag}.csv")
val_perclass_metrics_level2["model"] = "DeepLoc2-ProtT5"
val_perclass_level2.append(val_perclass_metrics_level2)
val_perclass_metrics_level3 = pd.read_csv(f"{dl2_path}/val_metrics_perclass_level3{tag}.csv")
val_perclass_metrics_level3["model"] = "DeepLoc2-ProtT5"
val_perclass_level3.append(val_perclass_metrics_level3)
```

```python
#Reconfigure LAProtT5 results
cols = ["acc", "macro_ap", "micro_ap", "f1_macro", "f1_micro", "jaccard_macro", "jaccard_micro", "rocauc_macro", "rocauc_micro", "mlrap", "cov_error", "num_labels", "level", "model"]

if SINGLE:
    tag="_single"
elif MULTI:
    tag="_multi"
else:
    tag = ""

test_avg_metrics = pd.read_csv(f"{LAPROTT5_OUTPUT}/LAProtT5_avg_metrics{tag}.csv")
test_avg_metrics.rename({"coverage_error": "cov_error"}, axis=1, inplace=True)
test_avg_metrics.loc[:, "level"] = test_avg_metrics.category_level.apply(lambda x: str(x[-1]))
test_avg_metrics["model"] = "LAProtT5"
test_avg_metrics = test_avg_metrics[cols]
test_avgs.append(test_avg_metrics)
cols = ["mcc_perclass", "acc_perclass", "recall_perclass", "precision_perclass", "f1_perclass", "jaccard_perclass", "rocauc_perclass", "label", "model"]
test_perclass_metrics_level1 = pd.read_csv(f"{LAPROTT5_OUTPUT}/LAProtT5_perclass_metrics_level1{tag}.csv")
test_perclass_metrics_level1.rename(
    {"category": "label",
    "location": "label",
    "mcc": "mcc_perclass",
    "acc": "acc_perclass",
    "recall": "recall_perclass",
    "precision": "precision_perclass",
    "f1": "f1_perclass",
    "jaccard": "jaccard_perclass",
    "rocauc": "rocauc_perclass"}, axis=1, inplace=True)
test_perclass_metrics_level1["model"] = "LAProtT5"
test_perclass_metrics_level1 = test_perclass_metrics_level1[cols]
test_perclass_level1.append(test_perclass_metrics_level1)
test_perclass_metrics_level2 = pd.read_csv(f"{LAPROTT5_OUTPUT}/LAProtT5_perclass_metrics_level2{tag}.csv")
test_perclass_metrics_level2.rename(
    {"category": "label",
    "location": "label",
    "mcc": "mcc_perclass",
    "acc": "acc_perclass",
    "recall": "recall_perclass",
    "precision": "precision_perclass",
    "f1": "f1_perclass",
    "jaccard": "jaccard_perclass",
    "rocauc": "rocauc_perclass"}, axis=1, inplace=True)
test_perclass_metrics_level2["model"] = "LAProtT5"
test_perclass_metrics_level2 = test_perclass_metrics_level2[cols]
test_perclass_level2.append(test_perclass_metrics_level2)
test_perclass_metrics_level3 = pd.read_csv(f"{LAPROTT5_OUTPUT}/LAProtT5_perclass_metrics_level3{tag}.csv")
test_perclass_metrics_level3.rename(
    {"category": "label",
    "location": "label",
    "mcc": "mcc_perclass",
    "acc": "acc_perclass",
    "recall": "recall_perclass",
    "precision": "precision_perclass",
    "f1": "f1_perclass",
    "jaccard": "jaccard_perclass",
    "rocauc": "rocauc_perclass"}, axis=1, inplace=True)
test_perclass_metrics_level3["model"] = "LAProtT5"
test_perclass_metrics_level3 = test_perclass_metrics_level3[cols]
test_perclass_level3.append(test_perclass_metrics_level3)
```

```python
#Collect results for random baseline
if SINGLE:
    tag="_single"
elif MULTI:
    tag="_multi"
else:
    tag = ""

test_avg_metrics = pd.read_csv(f"{RANDOM_OUTPUT}/random_avg_metrics{tag}.csv")
test_avg_metrics["model"] = "Random"
test_avgs.append(test_avg_metrics)
test_perclass_metrics_level1 = pd.read_csv(f"{RANDOM_OUTPUT}/random_perclass_metrics_level1{tag}.csv")
test_perclass_metrics_level1["model"] = "Random"
test_perclass_metrics_level1.rename({"labels": "label"}, axis=1, inplace=True)
test_perclass_level1.append(test_perclass_metrics_level1)
test_perclass_metrics_level2 = pd.read_csv(f"{RANDOM_OUTPUT}/random_perclass_metrics_level2{tag}.csv")
test_perclass_metrics_level2["model"] = "Random"
test_perclass_metrics_level2.rename({"labels": "label"}, axis=1, inplace=True)
test_perclass_level2.append(test_perclass_metrics_level2)
test_perclass_metrics_level3 = pd.read_csv(f"{RANDOM_OUTPUT}/random_perclass_metrics_level3{tag}.csv")
test_perclass_metrics_level3["model"] = "Random"
test_perclass_metrics_level3.rename({"labels": "label"}, axis=1, inplace=True)
test_perclass_level3.append(test_perclass_metrics_level3)
```

```python
test_avgs = pd.concat(test_avgs)
test_perclass_level1 = pd.concat(test_perclass_level1)
test_perclass_level2 = pd.concat(test_perclass_level2)
test_perclass_level3 = pd.concat(test_perclass_level3)

val_avgs = pd.concat(val_avgs)
val_perclass_level1 = pd.concat(val_perclass_level1)
val_perclass_level2 = pd.concat(val_perclass_level2)
val_perclass_level3 = pd.concat(val_perclass_level3)
```
